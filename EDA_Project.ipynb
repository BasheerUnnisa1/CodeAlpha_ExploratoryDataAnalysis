```python
# Importing Required Libraries

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Professional Chart Styling

sns.set_style("whitegrid")
plt.rcParams['figure.figsize'] = (12,6)

print("Libraries Imported Successfully")
```

```python
# Loading Dataset

df = pd.read_csv("netflix_titles.csv")

# Display First 5 Rows

df.head()
```

```python
# Dataset Information

print("Shape of Dataset:", df.shape)

print("\nColumn Names:\n")
print(df.columns)

print("\nDataset Information:\n")
print(df.info())
```

```python
# Checking Missing Values

missing_values = df.isnull().sum()

missing_values
```

```python
# Visualizing Missing Values

missing_values = df.isnull().sum()

missing_values = missing_values[missing_values > 0]

plt.figure(figsize=(10,5))

sns.barplot(
    x=missing_values.index,
    y=missing_values.values
)

plt.title("Missing Values in Netflix Dataset", fontsize=16)
plt.xlabel("Columns", fontsize=12)
plt.ylabel("Number of Missing Values", fontsize=12)

plt.xticks(rotation=45)

plt.show()
```

```python
# Movies vs TV Shows Distribution

plt.figure(figsize=(8,5))

sns.countplot(
    data=df,
    x='type',
    palette='Set2'
)

plt.title("Distribution of Movies and TV Shows on Netflix", fontsize=16)
plt.xlabel("Content Type", fontsize=12)
plt.ylabel("Count", fontsize=12)
plt.savefig("images/movies_vs_tvshows.png")
plt.show()
```

```python
# Top 10 Countries on Netflix

top_countries = df['country'].value_counts().head(10)

plt.figure(figsize=(12,6))

sns.barplot(
    x=top_countries.values,
    y=top_countries.index,
    palette='viridis'
)

plt.title("Top 10 Countries Producing Netflix Content", fontsize=16)
plt.xlabel("Number of Shows", fontsize=12)
plt.ylabel("Country", fontsize=12)
plt.savefig("images/top_countries.png")
plt.show()
```

```python
# Content Released Over The Years

release_year_count = df['release_year'].value_counts().sort_index()

plt.figure(figsize=(14,6))

sns.lineplot(
    x=release_year_count.index,
    y=release_year_count.values
)

plt.title("Netflix Content Release Trend Over Years", fontsize=16)
plt.xlabel("Release Year", fontsize=12)
plt.ylabel("Number of Releases", fontsize=12)
plt.savefig("images/release_trend.png")
plt.show()
```

```python
# Most Common Ratings on Netflix

plt.figure(figsize=(12,6))

sns.countplot(
    data=df,
    y='rating',
    order=df['rating'].value_counts().index,
    palette='coolwarm'
)

plt.title("Most Common Ratings on Netflix", fontsize=16)
plt.xlabel("Count", fontsize=12)
plt.ylabel("Rating", fontsize=12)
plt.savefig("images/ratings.png")
plt.show()
```

```python
# Top 10 Genres on Netflix

top_genres = df['listed_in'].value_counts().head(10)

plt.figure(figsize=(12,6))

sns.barplot(
    x=top_genres.values,
    y=top_genres.index,
    palette='magma'
)

plt.title("Top 10 Genres on Netflix", fontsize=16)
plt.xlabel("Number of Shows", fontsize=12)
plt.ylabel("Genres", fontsize=12)
plt.savefig("images/genres.png")
plt.show()
```

```python
# Extracting Movie Duration

movies = df[df['type'] == 'Movie'].copy()

movies['duration'] = movies['duration'].str.replace(' min', '')

movies['duration'] = pd.to_numeric(movies['duration'], errors='coerce')

# Plotting Movie Duration Distribution

plt.figure(figsize=(12,6))

sns.histplot(
    movies['duration'],
    bins=30,
    kde=True
)

plt.title("Distribution of Movie Durations on Netflix", fontsize=16)
plt.xlabel("Duration (Minutes)", fontsize=12)
plt.ylabel("Number of Movies", fontsize=12)
plt.savefig("images/movie_duration.png")
plt.show()
```

```python
# Heatmap for Missing Values

plt.figure(figsize=(12,6))

sns.heatmap(
    df.isnull(),
    cbar=False,
    cmap='viridis'
)

plt.title("Missing Values Heatmap", fontsize=16)

plt.show()
```
